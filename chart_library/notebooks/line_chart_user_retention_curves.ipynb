## Код для построения

```python
# =====================================================
# IMPORTS
# =====================================================

import pandas as pd
import matplotlib.pyplot as plt

# =====================================================
# LOAD DATA
# =====================================================

user_retention = pd.read_csv('user_retention.csv', index_col=0)

# =====================================================
# USER RETENTION CURVES
# =====================================================

plt.figure(figsize=(10,6))

for cohort in user_retention.index:

    plt.plot(
        user_retention.columns,
        user_retention.loc[cohort],
        marker='o',
        label=str(cohort),
        alpha=0.8
    )

plt.title('User Retention Curves', fontsize=14)
plt.xlabel('Months Since Acquisition', fontsize=12)
plt.ylabel('Retention Rate', fontsize=12)

plt.legend(
    title='Cohort',
    bbox_to_anchor=(1.05,1),
    loc='upper left'
)

plt.legend(loc='best')

plt.grid(
    True,
    axis='y', 
    alpha=0.3,
    linestyle='--'
)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

#plt.savefig("user_retention_curves.png", dpi=300, bbox_inches="tight")
plt.show()
```
